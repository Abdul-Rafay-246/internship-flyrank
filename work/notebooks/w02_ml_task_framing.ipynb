## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

csv_path = "../../data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(csv_path)

lane_data = df[(df["impressions_90d"] >=  100)
               & (df["sessions_90d"] > 0)].copy()

task_type = "Ranking"

print ("ML task type:", task_type)
print ("Pages in the lane 2 data:", len(lane_data))
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

# 0 mean the page is not marked as declining
lane_data["declining_proxy"] = 0

# 1 means the page is marked as declining
lane_data.loc[
    lane_data["trend_direction"] == "down",
    "declining_proxy"
] = 1

declining_pages = lane_data["declining_proxy"].sum()
proxy_rate = lane_data["declining_proxy"].mean() * 100

print("Pages with proxy value 1:", declining_pages)
print("Pages with proxy value 0:", len(lane_data) - declining_pages)
print("Declining proxy rate:", round(proxy_rate, 1), "%")
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
metric_name = "Precision@50"
pages_reviewed = 50
good_precision = 0.70

correct_pages_needed = int(pages_reviewed * good_precision)

print("Success metric:", metric_name)
print("Pages reviewed:", pages_reviewed)
print("Goal:", correct_pages_needed, "or more useful pages in the top 50")
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
columns_to_show = [
    "content_id",
    "content_type",
    "content_age_days",
    "days_since_last_update",
    "impressions_90d",
    "ctr",
    "avg_position",
    "engagement_rate",
    "declining_proxy"
]

unit_dataframe = lane_data[columns_to_show].copy()

print("Number of rows:", len(unit_dataframe))
print("Number of unique pages:", unit_dataframe["content_id"].nunique())

unit_dataframe.head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
stale_rule = (
    (lane_data["days_since_last_update"] >= 180) &
    (lane_data["impressions_90d"] >= 500)
)

low_ctr_rule = (
    (lane_data["impressions_90d"] >= 500) &
    (lane_data["avg_position"] > 0) &
    (lane_data["avg_position"] <= 20) &
    (lane_data["ctr"] < 0.5)
)

low_engagement_rule = (
    (lane_data["sessions_90d"] >= 30) &
    (
        (lane_data["engagement_rate"] < 30) |
        (lane_data["scroll_rate"] < 30)
    )
)

number_of_signals = (
    stale_rule.astype(int) +
    low_ctr_rule.astype(int) +
    low_engagement_rule.astype(int)
)

print("Pages selected by the stale rule:", stale_rule.sum())
print("Pages selected by the low-CTR rule:", low_ctr_rule.sum())
print("Pages selected by the low-engagement rule:", low_engagement_rule.sum())
print("Pages matching two or more signals:", (number_of_signals >= 2).sum())
```
